```python
import os
import getpass
import subprocess
from dotenv import load_dotenv
from google.colab import files
import shutil

# --- Ask the user for their preferred method ---
print("Please choose how to add your secret environment variables:")
print("  (1) Manually (most secure, but slow)")
print("  (2) Upload a .env file (fast, secure for this session)")

choice = input("Enter your choice (1 or 2): ").strip()

if choice == '1':
    # --- METHOD 1: MANUAL ENTRY ---
    print("\n--- Manual Entry ---")
    print("Please enter your secrets. The input will be hidden.")
    
    os.environ['API_ID'] = getpass.getpass('Enter your API_ID: ')
    os.environ['API_HASH'] = getpass.getpass('Enter your API_HASH: ')
    os.environ['BOT_TOKEN'] = getpass.getpass('Enter your BOT_TOKEN: ')
    os.environ['ADMIN'] = getpass.getpass('Enter your ADMIN User ID: ')
    os.environ['DATABASE_URL'] = getpass.getpass('Enter your DATABASE_URL: ')
    os.environ['LOG_CHANNEL'] = getpass.getpass('Enter your LOG_CHANNEL ID: ')
    
    # Optional Vars
    print("\n--- Optional Secrets (press Enter to skip) ---")
    os.environ['DATABASE_NAME'] = getpass.getpass('Enter your DATABASE_NAME (optional): ')
    os.environ['FORCE_SUBS'] = getpass.getpass('Enter your FORCE_SUBS ID (optional): ')
    os.environ['STRING_SESSION'] = getpass.getpass('Enter your STRING_SESSION (optional): ')
    os.environ['GEMINI_API_KEY'] = getpass.getpass('Enter your GEMINI_API_KEY (optional): ')
    
    print("\n✅ All secrets set manually.")

elif choice == '2':
    # --- METHOD 2: UPLOAD .ENV FILE ---
    print("\n--- .env Upload ---")
    
    # 1. Install the required library
    print("Installing python-dotenv...")
    subprocess.run(['pip', 'install', 'python-dotenv'], capture_output=True)

    # 2. Prompt user to upload
    print("Please upload your .env file. It will be deleted when this session ends.")
    uploaded = files.upload()
    
    if not uploaded:
        print("❌ No file uploaded. Please re-run the cell and try again.")
    else:
        # 3. Get the name of the uploaded file
        filename = list(uploaded.keys())[0]

        # Rename the uploaded file to exactly '.env' so bot.py can find it if needed
        if filename != '.env':
            shutil.move(filename, '.env')
            filename = '.env'
        
        # 4. Load the file into the environment
        load_dotenv(filename)
        
        print(f"\n✅ Successfully loaded secrets from {filename}.")
        
        # 5. (Optional but good) Verify some keys were loaded
        if os.environ.get('BOT_TOKEN'):
            print("   -> BOT_TOKEN found!")
        if os.environ.get('DATABASE_URL'):
            print("   -> DATABASE_URL found!")

else:
    print("❌ Invalid choice. Please re-run the cell and enter '1' or '2'.")
```
